```python
### Directory Loader
from langchain_community.document_loaders import DirectoryLoader

## load all the text files from the directory
dir_loader=DirectoryLoader(
    "../data/textfiles",
    glob="**/*.txt", ## Pattern to match files  
    loader_cls= TextLoader, ##loader class to use
    loader_kwargs={'encoding': 'utf-8'},
    show_progress=False

)

documents=dir_loader.load()
documents
```

```python
## Downloading PDF from url

import requests

# 1. Specify the PDF URL and target local filename
url = "https://www.publicissapient.com/content/dam/ps-reinvent/us/en/2025/04/insights-lp/agentic-ai-vs-generative-ai/doc/AgenticAI-vs-GenerativeAI-New.pdf"
filename = "agentic_ai_vs_genai.pdf"

# 2. Send an HTTP GET request to the URL
response = requests.get(url)

# 3. Check if the request was successful (Status Code 200)
if response.status_code == 200:
    # 4. Open a local file in 'write binary' (wb) mode and save the content
    with open(filename, "wb") as file:
        file.write(response.content)
    print(f"Successfully downloaded {filename}")
else:
    print(f"Failed to download. HTTP Status Code: {response.status_code}")
```
